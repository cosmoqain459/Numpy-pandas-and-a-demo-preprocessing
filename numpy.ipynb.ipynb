**Array reshaping**

```python
#Using the Reshape
arr = np.array([1,2,3,4,5,6])
reshaped = arr.reshape(2,3) # Converts the single row into 2 row and 3 coloumn
print(reshaped)

#Using ravel
#ravel converts any array to 1d
arr = np.array([[1,2,3],[4,5,6]])
ravelled = arr.ravel()
print(ravelled)

ravelled[0] = 100
print(ravelled)
print(arr)

#Using flatten
flat = arr.flatten()
print(flat)
flat[0] = 100
print(flat)
print(arr)
# *The key takeaway is that the changes applied to ravel affect original array whereas in flatten not as it returns the copy*



```

**Arithmetic operations**

```python
#Some simple arithematic operations
a  = np.array([1,2,3,4,5])
b= np.array([2,4,5,67,6])# The dimension should be same of both arrays as the operation applied is from element to element
print(a+b)
print(a-b)
print(a/b)
print(a*b)

#Universal functions-ufuncs
arr = np.array([1,4,9,16,25])
print(np.sqrt(arr))#prints the square root of the array elements

#Sine functions-np.sin
angles = np.array([0,np.pi,np.pi/2])# Here the valules are the angles of trignometry functions
print(np.sin(angles))
# Same for the other trignometry functions
print(np.cos(angles))
print(np.tan(angles))




```

**Iterating**

```python
#nditer() lets us iterate over any array
arr = np.array([[1,2],[3,4]])
for x in np.nditer(arr):
      print(x,end=" ")#prints in the same line while( " " ) gives the space between each elements
#ndenumerate() gives both index and value
arr = np.array([1,2,3,4,5])
for x in np.ndenumerate(arr):
      print(x)
```

**Views vs Copies**

```python
#view
arr = np.array([1,2,3,4,5])
view = arr[1:3]
view[0] = 200
view
arr

#copy
arr = np.array([1,2,3,4,5])
copy = arr[1:3].copy()
copy[0] = 200
copy
arr
#Original array is not changed in copy , while it is changed in the view
```

 ***Concatenation***

```python
#Concatenate
# Used for combining two arrays sidewise
a = np.array([1,2])
b = np.array([3,4])
combine =  np.concatenate((a,b))
combine

arr1 = np.array([[1,2],[3,4]])
arr2 = np.array([[5,6], [7,8]])
print(np.hstack((arr1,arr2)))# stacks array sideways
print(np.vstack((arr1,arr2)))# stacks array one upon another

# Other methods
print(np.stack((arr1,arr2),axis = 0))
print(np.stack((arr1,arr2),axis = 1))
```

**Repeation of the array**

**Aggregate function and Cumalative operations**


```python
arr = np.array([1,2,3])
print(np.sum(arr))
print(np.mean(arr))
print(np.median(arr))
print(np.min(arr))
print(np.max(arr))
print(np.std(arr))# prints the standard deviation of the elements in the array

#Cumalative operation
arr = np.array([1,2,3])
print(np.cumsum(arr))
print(np.cumprod(arr))

# Gives cumalitive sum [ 1st element ,1st element+2nd element,1st element+2nd element+3rd element]
# Gives cumalitive product [  1stelement ,1st element*2nd element,1st element*2nd element*3rd element]
```

**Where and agwhere**

```python
#Where
arr = np.array([1,2,3])
result = np.where(arr<2,'low','high')
result
#prints low when the condition is met and high when it's not.

#Argwhere
arr = np.array([10,25,30,50])
print(np.argwhere(arr>20))
#gives the indexes where the condition is met

```

**Advanve condition operations**

```python
#logical_and
arr = np.array([100,2,3,4,5,6])
mask = np.logical_and(arr>3,arr<6)# sets the boundary
print(mask)#prints the true and false
arr[mask]# prints the elements which meet the criteria

#logical_or
arr = np.array([100,2,3,4,5,6])
mask = np.logical_orr(arr>3,arr<6)# sets the boundary
print(mask)#prints the true and false
arr[mask]# prints the elements which does not meet the criteria of logical_and(for understanding)


```

**Broadcasting and vectorization**

```python
#BROADCASTING
# To make the smaller array stretch to match the bigger array
# condition- One array has to be of 1D
image = np.array([[200,500],[250,100]])
brightness = image + 50
brightness
# 50 is added to each element

#VECTORIZATION
#np.vectorize - converts a regular function to be applied on an array
def square(x):
   return x*x
arr = np.array([1,2,3,4,5,6])
vfunc = np.vectorize(square)
vfunc(arr)


```
