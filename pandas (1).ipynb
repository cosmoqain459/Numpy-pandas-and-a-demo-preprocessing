```python
import pandas as pd

# Recreate the DataFrame as it was before the drop operations
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Ages alive': [24, 27, 22, 32, 29], # 'Age' was renamed to 'Ages alive'
    'City': ['Delhi', 'Mumbai', 'Kolkata', 'Chennai', 'Pune'],
    'Score': [85, 92, 78, 88, 95]
}

df = pd.DataFrame(data)
df['Spouse'] = [2, 3, 4, 5, 2] # 'Spouse' column was added
df = df.rename(index={2:'two'}) # Index 2 was renamed to 'two'
df['Country'] = 'India' # 'Country' column was added

# Now, perform the drop operations
df = df.drop('Country', axis=1) # Drop the 'Country' column
df = df.drop('two', axis=0)      # Drop the row with index 'two'

print(df)
```
